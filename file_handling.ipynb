# 1. Create a student record system that allows adding student data (name, roll, marks) and saving the data to a file.

```python
file = open("students.txt", "a")

name = input("Enter Name: ")
roll = input("Enter Roll No: ")
marks = input("Enter Marks: ")

file.write(name + "," + roll + "," + marks + "\n")
file.close()

print("Student record added successfully.")
```

# 2. Implement search functionality to find a student by roll number.

```python
roll = input("Enter Roll No to search: ")

file = open("students.txt", "r")
found = False

for line in file:
    data = line.strip().split(",")
    if data[1] == roll:
        print("Name:", data[0])
        print("Roll:", data[1])
        print("Marks:", data[2])
        found = True
        break

if not found:
    print("Student not found.")

file.close()
```

# 5. Show all student records sorted by marks.

```python
file = open("students.txt", "r")
students = []

for line in file:
    data = line.strip().split(",")
    students.append(data)

file.close()

students.sort(key=lambda x: int(x[2]), reverse=True)

print("Name\tRoll\tMarks")
for s in students:
    print(s[0], "\t", s[1], "\t", s[2])
```

# 6. Display student records with marks greater than 75.

```python
file = open("students.txt", "r")

print("Students scoring more than 75:")

for line in file:
    data = line.strip().split(",")
    if int(data[2]) > 75:
        print(data[0], data[1], data[2])

file.close()
```

# 7. Export top 3 performing students to a new file.

```python
file = open("students.txt", "r")
students = []

for line in file:
    data = line.strip().split(",")
    students.append(data)

file.close()

students.sort(key=lambda x: int(x[2]), reverse=True)

out = open("top3.txt", "w")

for s in students[:3]:
    out.write(",".join(s) + "\n")

out.close()

print("Top 3 students exported successfully.")
```

# 8. Implement data backup feature (copying all data to backup.txt).

```python
source = open("students.txt", "r")
data = source.read()
source.close()

backup = open("backup.txt", "w")
backup.write(data)
backup.close()

print("Backup created successfully.")
```

# 9. Count how many students scored more than 90 and less than 35.

```python
file = open("students.txt", "r")

above90 = 0
below35 = 0

for line in file:
    data = line.strip().split(",")
    marks = int(data[2])

    if marks > 90:
        above90 += 1
    if marks < 35:
        below35 += 1

file.close()

print("Students scoring above 90:", above90)
print("Students scoring below 35:", below35)
```

# 10. Create a menu-driven system (Add, View, Search, Update, Delete, Exit).

```python
while True:
    print("\n----- Student Record System -----")
    print("1. Add")
    print("2. View")
    print("3. Search")
    print("4. Update")
    print("5. Delete")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        file = open("students.txt", "a")
        name = input("Enter Name: ")
        roll = input("Enter Roll No: ")
        marks = input("Enter Marks: ")
        file.write(name + "," + roll + "," + marks + "\n")
        file.close()
        print("Record Added.")

    elif choice == "2":
        file = open("students.txt", "r")
        print(file.read())
        file.close()

    elif choice == "3":
        roll = input("Enter Roll No: ")
        file = open("students.txt", "r")
        found = False
        for line in file:
            data = line.strip().split(",")
            if data[1] == roll:
                print("Name:", data[0], "Marks:", data[2])
                found = True
        if not found:
            print("Record not found.")
        file.close()

    elif choice == "4":
        
        roll = input("Enter Roll No to update: ")
        new_marks = input("Enter New Marks: ")

        file = open("students.txt", "r")
        lines = file.readlines()
        file.close()

        file = open("students.txt", "w")

        found = False

        for line in lines:
            data = line.strip().split(",")

            if data[1] == roll:
                data[2] = new_marks
                line = ",".join(data) + "\n"
                found = True

            file.write(line)

        file.close()

        if found:
            print("Marks updated successfully.")
        else:
            print("Student not found.")

    elif choice == "5":
        value = input("Enter Name or Roll No to delete: ")

        file = open("students.txt", "r")
        lines = file.readlines()
        file.close()

        file = open("students.txt", "w")

        found = False

        for line in lines:
            data = line.strip().split(",")

            if data[0] == value or data[1] == value:
                found = True
            else:
                file.write(line)

        file.close()

        if found:
            print("Record deleted successfully.")
        else:
            print("Record not found.")

    elif choice == "6":
            print("Thank You!")
            break

    else:
            print("Invalid Choice.")
```
